**ANOVA: Yield Across Seasons**

```python
from scipy.stats import f_oneway
```

```python
kharif_yield = df_clean[
    df_clean['Season'] == 'Kharif'
]['Yield_Tonnes_Ha']

rabi_yield = df_clean[
    df_clean['Season'] == 'Rabi'
]['Yield_Tonnes_Ha']

zaid_yield = df_clean[
    df_clean['Season'] == 'Zaid'
]['Yield_Tonnes_Ha']
```

```python
f_stat, p_value = f_oneway(
    kharif_yield,
    rabi_yield,
    zaid_yield
)

print("One-Way ANOVA — Yield Across Seasons")
print("-------------------------------------")
print("F-statistic:", round(f_stat, 4))
print("p-value:", round(p_value, 6))
```

**Interpret the ANOVA**

```python
alpha = 0.05

if p_value < alpha:
    print("Result: Statistically significant difference in mean yield across seasons.")
else:
    print("Result: No statistically significant difference in mean yield across seasons.")
```

**ANOVA for Profit**

```python
kharif_profit = df_clean[
    df_clean['Season'] == 'Kharif'
]['Profit_INR']

rabi_profit = df_clean[
    df_clean['Season'] == 'Rabi'
]['Profit_INR']

zaid_profit = df_clean[
    df_clean['Season'] == 'Zaid'
]['Profit_INR']
```

```python
profit_f_stat, profit_p_value = f_oneway(
    kharif_profit,
    rabi_profit,
    zaid_profit
)

print("One-Way ANOVA — Profit Across Seasons")
print("--------------------------------------")
print("F-statistic:", round(profit_f_stat, 4))
print("p-value:", round(profit_p_value, 6))
```

```python
if profit_p_value < 0.05:
    print("Result: Statistically significant difference in mean profit across seasons.")
else:
    print("Result: No statistically significant difference in mean profit across seasons.")
```

**Generate the final Key Findings automatically**

```python
# ============================================================
# FINAL KEY FINDINGS
# ============================================================

summary = df_clean.groupby('Season').agg(
    Avg_Yield=('Yield_Tonnes_Ha', 'mean'),
    Avg_Production=('Production_Tonnes', 'mean'),
    Avg_Revenue=('Revenue_INR', 'mean'),
    Avg_Cost=('Total_Cost_INR', 'mean'),
    Avg_Profit=('Profit_INR', 'mean'),
    Avg_Water=('Water_Used_m3', 'mean'),
    Avg_Water_Efficiency=('Water_Efficiency_t_per_1000m3', 'mean'),
    Avg_Disease_Risk=('Disease_Pest_Risk_pct', 'mean')
).round(2)

print("========== KEY FINDINGS ==========\n")

# Yield
highest_yield = summary['Avg_Yield'].idxmax()
lowest_yield = summary['Avg_Yield'].idxmin()

print(f"1. Highest average yield: {highest_yield} "
      f"({summary.loc[highest_yield, 'Avg_Yield']} tonnes/ha)")

print(f"   Lowest average yield: {lowest_yield} "
      f"({summary.loc[lowest_yield, 'Avg_Yield']} tonnes/ha)")

# Production
highest_production = summary['Avg_Production'].idxmax()

print(f"\n2. Highest average production: {highest_production} "
      f"({summary.loc[highest_production, 'Avg_Production']} tonnes)")

# Profit
highest_profit = summary['Avg_Profit'].idxmax()
lowest_profit = summary['Avg_Profit'].idxmin()

print(f"\n3. Highest average profit: {highest_profit} "
      f"(₹{summary.loc[highest_profit, 'Avg_Profit']:,.2f})")

print(f"   Lowest average profit: {lowest_profit} "
      f"(₹{summary.loc[lowest_profit, 'Avg_Profit']:,.2f})")

# Water
highest_water = summary['Avg_Water'].idxmax()
highest_efficiency = summary['Avg_Water_Efficiency'].idxmax()

print(f"\n4. Highest average water usage: {highest_water} "
      f"({summary.loc[highest_water, 'Avg_Water']:,.2f} m³)")

print(f"5. Highest average water efficiency: {highest_efficiency} "
      f"({summary.loc[highest_efficiency, 'Avg_Water_Efficiency']:.2f} tonnes/1000 m³)")

# Risk
highest_risk = summary['Avg_Disease_Risk'].idxmax()

print(f"\n6. Highest average disease/pest risk: {highest_risk} "
      f"({summary.loc[highest_risk, 'Avg_Disease_Risk']:.2f}%)")

# Statistical findings
print("\n========== STATISTICAL FINDINGS ==========")

print(f"\nYield ANOVA p-value: {p_value:.6f}")

if p_value < 0.05:
    print("Yield: Significant seasonal difference detected.")
else:
    print("Yield: No statistically significant seasonal difference detected.")

print(f"\nProfit ANOVA p-value: {profit_p_value:.6f}")

if profit_p_value < 0.05:
    print("Profit: Significant seasonal difference detected.")
else:
    print("Profit: No statistically significant seasonal difference detected.")
```
